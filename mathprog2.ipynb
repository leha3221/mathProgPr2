# Практическое занятие № 2. Комплексные числа и их геометрия

**Цель работы:** освоить алгебраическую, тригонометрическую и показательную формы комплексных чисел; вычислять корни и решать уравнения над \(\mathbb C\); изображать геометрические множества и строить множества Жюлиа и кривые на сфере Римана.

## Краткая теория

Комплексное число записывается как \(z=x+iy\), где \(x=\operatorname{Re}z\), \(y=\operatorname{Im}z\), \(i^2=-1\). Модуль \(r=|z|=\sqrt{x^2+y^2}\), аргумент \(\varphi=\arg z\). Три формы:
- **алгебраическая:** \(z=x+iy\);
- **тригонометрическая:** \(z=r(\cos\varphi+i\sin\varphi)\);
- **показательная:** \(z=re^{i\varphi}\).

При умножении модули перемножаются, аргументы складываются; при делении модули делятся, аргументы вычитаются. Формула Муавра:
\[
(re^{i\varphi})^n=r^n e^{in\varphi}.
\]
Все \(n\)-е корни числа \(re^{i\varphi}\):
\[
z_k=r^{1/n}e^{i(\varphi+2\pi k)/n},\quad k=0,\ldots,n-1.
\]
Геометрический смысл: \(|z-a|\) — расстояние от точки \(z\) до \(a\); \(\operatorname{Re}z=x\), \(\operatorname{Im}z=y\). Для вычислений используются `cmath`, `numpy`, `scipy.special.gamma`; для графиков — `matplotlib`.

**Примечание:** в заданиях 2.4(e) и 2.9 формулы в исходном PDF частично неоднозначны/необычны. Ниже сохранена читаемая запись, а неоднозначности отмечены в соответствующих разделах.

```python
import numpy as np
import matplotlib.pyplot as plt
import cmath
from scipy.special import gamma
from scipy.optimize import root
from mpl_toolkits.mplot3d import Axes3D

np.set_printoptions(precision=6, suppress=True)
plt.rcParams['figure.figsize'] = (7, 5)
plt.rcParams['axes.grid'] = True

def show_complex(z, name='z'):
    r, phi = abs(z), cmath.phase(z)
    print(f'{name} = {z.real:.8g} {z.imag:+.8g}i')
    print(f'  алгебраическая форма: {z.real:.8g} {z.imag:+.8g}i')
    print(f'  тригонометрическая форма: {r:.8g}(cos({phi:.8g}) + i sin({phi:.8g}))')
    print(f'  показательная форма: {r:.8g} * exp(i*{phi:.8g})')
```

```python
# a) ((5+i)(7-6i))/(3+i)
z_a = (5+1j)*(7-6j)/(3+1j)
show_complex(z_a, 'a')

# b) (1+i)^5 / (1-i)^3
z_b = (1+1j)**5/(1-1j)**3
show_complex(z_b, 'b')

# c) (-1+i*sqrt(3))^15/(1-i)^20 + (-1-i*sqrt(3))^15/(1+i)^20
z_c = ((-1+1j*np.sqrt(3))**15/(1-1j)**20
       + (-1-1j*np.sqrt(3))**15/(1+1j)**20)
show_complex(z_c, 'c')
```

## 2.2. Решить \(z^8=1+i\) и изобразить корни

```python
w = 1 + 1j
n = 8
roots_22 = [abs(w)**(1/n)*np.exp(1j*(np.angle(w)+2*np.pi*k)/n) for k in range(n)]
print('Корни уравнения z^8 = 1+i:')
for k, z in enumerate(roots_22):
    print(f'z_{k} = {z.real:.6f} {z.imag:+.6f}i; проверка z^8 = {z**8}')

fig, ax = plt.subplots()
ax.scatter(np.real(roots_22), np.imag(roots_22))
for k,z in enumerate(roots_22):
    ax.annotate(f'z{k}', (z.real,z.imag), xytext=(5,5), textcoords='offset points')
ax.axhline(0,color='black',lw=.7); ax.axvline(0,color='black',lw=.7)
ax.set_aspect('equal'); ax.set_xlabel('Re z'); ax.set_ylabel('Im z')
ax.set_title('Корни уравнения z⁸ = 1+i')
plt.show()
```

## 2.3. Вычислить корни над \(\mathbb C\) и изобразить ответы

Используем формулу корней из теоретического блока. Для корней четвёртой, третьей и девятой степени строим все ветви, а не только главное значение.

```python
def nth_roots(w, n):
    # Все n корней комплексного числа w
    if abs(w) == 0:
        return [0j]*n
    return [abs(w)**(1/n)*np.exp(1j*(np.angle(w)+2*np.pi*k)/n) for k in range(n)]

# a) восьмой корень из 1
r23a = nth_roots(1+0j, 8)

# b) четвертый корень из -18/(1+i*sqrt(3))
r23b = nth_roots(-18/(1+1j*np.sqrt(3)), 4)

# c) кубический корень из (1-5i)/(1+i) - 5(1+2i)/(2-i) + 11
inside_c = (1-5j)/(1+1j) - 5*(1+2j)/(2-1j) + 11
r23c = nth_roots(inside_c, 3)

# d) девятый корень из 1-i
r23d = nth_roots(1-1j, 9)

for label, roots in [('a',r23a), ('b',r23b), ('c',r23c), ('d',r23d)]:
    print(f'\n{label}:')
    for z in roots:
        print(f'{z.real:.6f} {z.imag:+.6f}i')

fig, axs = plt.subplots(2,2,figsize=(10,9))
for ax,(label,roots) in zip(axs.flat,[('a) ⁸√1',r23a),('b) ⁴√(-18/(1+i√3))',r23b),
                                      ('c) кубические корни',r23c),('d) ⁹√(1-i)',r23d)]):
    ax.scatter(np.real(roots),np.imag(roots))
    ax.axhline(0,color='black',lw=.6); ax.axvline(0,color='black',lw=.6)
    ax.set_aspect('equal'); ax.set_title(label); ax.set_xlabel('Re'); ax.set_ylabel('Im')
plt.tight_layout(); plt.show()
```

## 2.4. Геометрические множества на комплексной плоскости

Положим \(z=x+iy\). Условия с модулем интерпретируются как расстояния, а условия на аргумент задают угловые области. Для аргументных неравенств ниже используется principal argument в диапазоне \((-\pi,\pi]\).

```python
# a) |z-i| + |z+i| < 4: внутренняя область эллипса с фокусами (0,±1)
x = np.linspace(-2.2,2.2,500); y = np.linspace(-2.2,2.2,500)
X,Y = np.meshgrid(x,y); Z = X+1j*Y
fig, axs = plt.subplots(2,3,figsize=(14,8))
axs[0,0].contourf(X,Y,(np.abs(Z-1j)+np.abs(Z+1j))<4,levels=[.5,1.5],alpha=.6)
axs[0,0].set_title('2.4(a): |z-i|+|z+i|<4')

# b) Re(z(1-i)) = x+y < sqrt(2)
axs[0,1].contourf(X,Y,X+Y<np.sqrt(2),levels=[.5,1.5],alpha=.6)
axs[0,1].plot(x,np.sqrt(2)-x,'r'); axs[0,1].set_title('2.4(b): x+y<√2')

# c) equidistant from -1, -i, i
axs[0,2].scatter([-1,0,0],[0,-1,1],label='заданные точки')
axs[0,2].scatter([0],[0],marker='x',s=100,label='пересечение серединных перпендикуляров')
axs[0,2].set_title('2.4(c): |z+1|=|z+i|=|z-i|')
axs[0,2].legend(fontsize=7)

# d) 0 < arg((i-z)/(z+i)) < pi/2
Qd=(1j-Z)/(Z+1j)
maskd=(np.angle(Qd)>0)&(np.angle(Qd)<np.pi/2)&(np.abs(Z+1j)>1e-3)
axs[1,0].contourf(X,Y,maskd,levels=[.5,1.5],alpha=.6)
axs[1,0].set_title('2.4(d): аргумент в (0, π/2)')

# e) Source states -2π/3 <= arg((2-iz)/(z+3i)) <= -5π/6.
# The written bounds are reversed numerically, so show the literal inequality as an empty region.
axs[1,1].text(.05,.5,'Границы в PDF записаны в обратном порядке:
−2π/3 ≤ arg(...) ≤ −5π/6.
При обычном сравнении интервал пуст.',transform=axs[1,1].transAxes,va='center')
axs[1,1].set_title('2.4(e): замечание к условию')

# f) -2pi/3 <= arg((3-z)/(z+i)) <= -pi/3
Qf=(3-Z)/(Z+1j)
angf=np.angle(Qf)
maskf=(angf>=-2*np.pi/3)&(angf<=-np.pi/3)&(np.abs(Z+1j)>1e-3)
axs[1,2].contourf(X,Y,maskf,levels=[.5,1.5],alpha=.6)
axs[1,2].set_title('2.4(f): угловая область')

for ax in axs.flat:
    ax.axhline(0,color='black',lw=.5); ax.axvline(0,color='black',lw=.5)
    ax.set_xlim(-2.2,2.2); ax.set_ylim(-2.2,2.2); ax.set_aspect('equal')
    ax.set_xlabel('x = Re z'); ax.set_ylabel('y = Im z')
plt.tight_layout(); plt.show()

print('2.4(c): единственная точка, равноудалённая от -1, -i и i, — z=0.')
```

## 2.5. Доказать, что уравнение задаёт окружность

Дано \(A|z|^2+\overline Bz+B\overline z+C=0\), где \(A,C\in\mathbb R\), \(A>0\), \(AC<|B|^2\). После деления на \(A\) и выделения полного квадрата:
\[
\left|z+\frac{B}{A}\right|^2=\frac{|B|^2-AC}{A^2}.
\]
Следовательно, центр окружности \(z_0=-B/A\), радиус \(R=\sqrt{|B|^2-AC}/A\). Условие \(AC<|B|^2\) обеспечивает \(R>0\).

```python
# Проверка на численном примере (параметры можно заменить)
A, C, B = 2.0, 1.0, 1+2j
assert A > 0 and A*C < abs(B)**2
center = -B/A
radius = np.sqrt(abs(B)**2-A*C)/A
print('Центр:', center)
print('Радиус:', radius)
t=np.linspace(0,2*np.pi,500)
circle=center+radius*np.exp(1j*t)
plt.plot(circle.real,circle.imag)
plt.scatter([center.real],[center.imag],marker='x',label='центр')
plt.axis('equal'); plt.xlabel('Re z'); plt.ylabel('Im z'); plt.title('Окружность из 2.5'); plt.legend(); plt.show()
```

## 2.7. Обобщение биномиальных коэффициентов (Jupyter Notebook)

Используем определение
\[
\binom zw=\frac{\Gamma(z+1)}{\Gamma(w+1)\Gamma(z-w+1)}.
\]
Функция `gamma` из SciPy позволяет вычислять это выражение для комплексных аргументов. Для устойчивости используем комплексную гамма-функцию из `scipy.special`.

```python
from scipy.special import loggamma

def generalized_binomial(z, w):
    # Эквивалент определения через Gamma, вычисление через loggamma устойчивее
    return np.exp(loggamma(z+1) - loggamma(w+1) - loggamma(z-w+1))

# a) C_(1+i)^(1-i), то есть верхний индекс w=1-i, нижний z=1+i
value_27 = generalized_binomial(1+1j, 1-1j)
print('C_(1+i)^(1-i) =', value_27)

# b) главный аргумент
print('Arg =', np.angle(value_27), 'радиан')
print('Arg =', np.degrees(np.angle(value_27)), 'градусов')
```

### 2.7(c). Линии уровня аргумента

В записи PDF обозначение линии уровня отображается неоднозначно. Ниже построены линии уровня \(\arg\binom{z}{1/z}=\mathrm{const}\), то есть интерпретация формулы как биномиального коэффициента с верхним параметром \(z\) и нижним \(1/z\). Полюса и численные переполнения маскируются.

```python
# Сетка ограничена, чтобы график был читаемым
xx = np.linspace(-3,3,500)
yy = np.linspace(-3,3,500)
XX,YY = np.meshgrid(xx,yy)
ZZ = XX+1j*YY
W = np.full(ZZ.shape, np.nan+1j*np.nan, dtype=complex)
mask = (np.abs(ZZ)>0.08)
# Вычисляем по точкам, пропуская особенности; это может занять некоторое время.
for idx in zip(*np.where(mask)):
    z = ZZ[idx]
    try:
        val = generalized_binomial(z, 1/z)
        if np.isfinite(val.real) and np.isfinite(val.imag) and abs(val)<1e8:
            W[idx] = val
    except (ValueError, OverflowError, ZeroDivisionError):
        pass
phase = np.angle(W)
plt.figure(figsize=(8,6))
plt.contour(XX,YY,phase,levels=np.linspace(-np.pi,np.pi,17),linewidths=.6)
plt.xlabel('Re z'); plt.ylabel('Im z'); plt.title(r'Линии уровня arg(C(z, 1/z))')
plt.axis('equal'); plt.show()
```

## 2.9. Множества Жюлиа

Для отображения \(f_c(z)=z^2+c\) итерации \(z_{n+1}=z_n^2+c\) расходятся к бесконечности, если модуль достаточно велик. На сетке комплексной плоскости считаем число итераций до выхода за радиус \(R\). Цвет показывает время выхода. Далее строим два множества для параметров, указанных в листке, а также множества для двух рациональных отображений, формулы которых читаются в PDF как \((iz^5-1)/(z^5-i)\) и \((z^2+i)/(1-z^2)\).

```python
def julia_escape(c, width=600, height=600, xmin=-2, xmax=2, ymin=-2, ymax=2, max_iter=250, radius=10, func=None):
    xs = np.linspace(xmin,xmax,width)
    ys = np.linspace(ymin,ymax,height)
    X,Y = np.meshgrid(xs,ys)
    Z = X+1j*Y
    escape = np.zeros(Z.shape, dtype=float)
    active = np.ones(Z.shape, dtype=bool)
    for n in range(max_iter):
        if func is None:
            Z[active] = Z[active]**2 + c
        else:
            Z[active] = func(Z[active])
        escaped_now = active & (np.abs(Z)>radius)
        escape[escaped_now] = n+1
        active[escaped_now] = False
    escape[active] = max_iter
    return escape

fig, axs = plt.subplots(1,2,figsize=(12,5))
for ax,c in zip(axs,[complex(-0.1,0.65),complex(0.28,0.009)]):
    img=julia_escape(c,max_iter=250)
    ax.imshow(img,origin='lower',extent=(-2,2,-2,2),cmap='magma')
    ax.set_title(f'c = {c.real} {c.imag:+}i'); ax.set_xlabel('Re z'); ax.set_ylabel('Im z')
plt.tight_layout(); plt.show()
```

```python
# Рациональные отображения из задания 2.9(d)
def f1(z):
    return (1j*z**5 - 1)/(z**5 - 1j)

def f2(z):
    return (z**2 + 1j)/(1-z**2)

fig, axs = plt.subplots(1,2,figsize=(12,5))
for ax,func,title in [(axs[0],f1,r'$(iz^5-1)/(z^5-i)$'),(axs[1],f2,r'$(z^2+i)/(1-z^2)$')]:
    img=julia_escape(0j,max_iter=180, radius=8, func=func)
    ax.imshow(img,origin='lower',extent=(-2,2,-2,2),cmap='turbo')
    ax.set_title(title); ax.set_xlabel('Re z'); ax.set_ylabel('Im z')
plt.tight_layout(); plt.show()
```

## 2.10. Кривые на сфере Римана

Стереографическая проекция отображает комплексную плоскость на единичную сферу:
\[
(x,y)\mapsto\left(\frac{2x}{1+x^2+y^2},\frac{2y}{1+x^2+y^2},
\frac{x^2+y^2-1}{1+x^2+y^2}\right).
\]
Точка на бесконечности соответствует северному полюсу сферы. На рисунках ниже на сферу переносятся прямая, окружность, парабола и гипербола.

```python
def stereographic(z):
    x, y = np.real(z), np.imag(z)
    r2 = x*x+y*y
    return 2*x/(1+r2), 2*y/(1+r2), (r2-1)/(1+r2)

def plot_on_riemann_sphere(ax, z, title):
    X,Y,Z = stereographic(z)
    u=np.linspace(0,2*np.pi,50); v=np.linspace(0,np.pi,30)
    sx=np.outer(np.cos(u),np.sin(v)); sy=np.outer(np.sin(u),np.sin(v)); sz=np.outer(np.ones_like(u),np.cos(v))
    ax.plot_surface(sx,sy,sz,color='gold',alpha=.16,linewidth=0)
    ax.plot(X,Y,Z,color='crimson',lw=2)
    ax.scatter([0],[0],[1],color='black',s=20)
    ax.set_title(title); ax.set_box_aspect((1,1,1))
    ax.set_xlim(-1,1); ax.set_ylim(-1,1); ax.set_zlim(-1,1)

t=np.linspace(-8,8,1600)
curves = [
    (t+1j*(0.5*t+0.5), 'Прямая y=0.5x+0.5'),
    (1.0+1.2*np.exp(1j*t), 'Окружность'),
    (t+1j*(0.25*t**2), 'Парабола y=x²/4'),
    (2/np.cos(t)+1j*np.tan(t), 'Гипербола x²/4-y²=1')
]
fig=plt.figure(figsize=(12,10))
for k,(z,title) in enumerate(curves,1):
    ax=fig.add_subplot(2,2,k,projection='3d')
    plot_on_riemann_sphere(ax,z,title)
plt.tight_layout(); plt.show()
```

### Логарифмическая спираль на сфере Римана

Спираль \(r=ae^{b\varphi}\) зададим параметрически: \(z(\varphi)=ae^{b\varphi}e^{i\varphi}\). При выбранных \(a=0.08\), \(b=0.12\) кривая визуализируется в комплексной плоскости и после стереографической проекции на сферу.

```python
phi=np.linspace(-8*np.pi,3*np.pi,3000)
a,b=0.08,0.12
r=a*np.exp(b*phi)
spiral=r*np.exp(1j*phi)
fig=plt.figure(figsize=(12,5))
ax1=fig.add_subplot(1,2,1)
ax1.plot(spiral.real,spiral.imag)
ax1.set_aspect('equal'); ax1.set_xlabel('Re z'); ax1.set_ylabel('Im z')
ax1.set_title('Логарифмическая спираль в комплексной плоскости')
ax2=fig.add_subplot(1,2,2,projection='3d')
plot_on_riemann_sphere(ax2,spiral,'Спираль на сфере Римана')
plt.tight_layout(); plt.show()
```
